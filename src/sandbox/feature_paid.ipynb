```python
import pandas as pd
```

```python
df_train = pd.read_parquet("../../data/churn-prediction-25-26/train.parquet")
```

```python
df = df_train.copy()
```

```python
df["last_time"] = df.groupby(["userId"])["time"].transform("max")
df["last_item"] = df.groupby(["userId", "sessionId"])["itemInSession"].transform("max")
```

```python
temp = df[(df["time"] == df["last_time"]) & \
          (df["itemInSession"] == df["last_item"])][["userId", "level"]]
```

```python
temp["level"] = temp["level"] == "paid"
temp["level"] = temp["level"].astype(int)
```
