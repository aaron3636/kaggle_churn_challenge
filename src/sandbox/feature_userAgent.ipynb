```python
import pandas as pd
from skrub import TableVectorizer, GapEncoder
```

```python
df_train = pd.read_parquet("../../data/churn-prediction-25-26/train.parquet")
```

```python
df = df_train.copy()
```

```python
df = df[["userId", "userAgent"]]
```

```python
df = df.drop_duplicates()
```

```python
df["device_used"] = df["userAgent"].apply(lambda x: x[x.index("("): x.index(")") + 1])
```

```python
def device_type(s: str):
    if "Windows" in s:
        return "Windows"
    elif "Macintosh" in s:
        return "Macintosh"
    elif "Linux" in s:
        return "Linux"
    elif "iPad" in s:
        return "iPad"
    elif "iPhone" in s:
        return "iPhone"
    else:
        return "Different Device"
```

```python
df["exact_device"] = df["device_used"].apply(device_type)
```
