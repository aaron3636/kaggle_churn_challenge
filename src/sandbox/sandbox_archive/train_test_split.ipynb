```python
import pandas as pd
```

```python
full_dataset_df = pd.read_parquet("../../data/churn-prediction-25-26/train.parquet")
```

```python
df_train = full_dataset_df.copy()
df_test = full_dataset_df.copy()
```

```python
# First Test set
#cutoff_train = pd.Timestamp("2018-10-08")
#cutoff_test = pd.Timestamp("2018-10-18")

# Second Train Test set
cutoff_train = pd.Timestamp("2018-10-15")
cutoff_test = pd.Timestamp("2018-10-25")

df_train = df_train[df_train["time"] <= cutoff_train]
df_test = df_test[df_test["time"] <= cutoff_test]
```

```python
# only consider users that haven't already cancelled their memebership
df_train["already_canceled"] = df_train["page"] == "Cancellation Confirmation"
df_train_reduced = df_train[~df_train["userId"].isin(df_train[df_train["already_canceled"]]["userId"])].copy()
```

```python
# in the test dataset: only consider users that are in the train dataset
# create an array of the form ["userId", "Cancellation Confirmed"]

df_test_reduced = df_test[df_test["userId"].isin(df_train_reduced["userId"])].copy()
df_test_reduced["Canceled"] = df_test_reduced["page"] == "Cancellation Confirmation"
df_test_reduced["Canceled"] = df_test_reduced["Canceled"].astype(int)
```

```python
df_test_reduced["Canceled"] = df_test_reduced.groupby("userId")["Canceled"].transform("max")
```

```python
df_test_reduced = df_test_reduced[["userId", "Canceled"]].drop_duplicates()
```
